```python
# IMPORTS
from transformers import pipeline
```

```python
# LOAD MODEL (FAST + STABLE)
generator = pipeline("text-generation", model="distilgpt2")
```

```python
print(generator("What is LangChain?", max_new_tokens=30)[0]["generated_text"])
```

```python
def prompt_template(topic):
    return f"Explain {topic} in simple terms."

print(generator(prompt_template("LangChain"), max_new_tokens=30)[0]["generated_text"])
```

```python
def chain(input_text):
    prompt = f"Explain clearly: {input_text}"
    return generator(prompt, max_new_tokens=30)[0]["generated_text"]

print(chain("Machine Learning"))
```

```python
chat_history = ""

def chat(input_text):
    global chat_history
    chat_history += f"\nUser: {input_text}"

    response = generator(chat_history, max_new_tokens=30)[0]["generated_text"]

    chat_history += f"\nAI: {response}"
    return response

print(chat("Hi, my name is V"))
print(chat("What is my name?"))
```

```python
def calculator(expression):
    return eval(expression)

print("Tool Output:", calculator("10 + 20"))
```
